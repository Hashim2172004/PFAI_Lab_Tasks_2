# Text Classification using Sklearn 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Sample dataset
texts = [
    "Ronaldo scored a goal in the last match",
    "The cricket team won the series",
    "Messi is the best football player",
    "The new iPhone was launched today",
    "AI is changing the technology world",
    "Samsung introduced a new processor"
]

labels = [
    "sports",
    "sports",
    "sports",
    "technology",
    "technology",
    "technology"
]

# Convert text → numeric features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.33, random_state=42)

# Train model
model = MultinomialNB()
model.fit(X_train, y_train)

# Predict
pred = model.predict(X_test)

# Accuracy
print("Predictions:", pred)
print("Actual:", y_test)
print("Accuracy:", accuracy_score(y_test, pred))
```
